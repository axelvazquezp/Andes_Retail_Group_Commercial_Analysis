### Diseño y planificación
✏️ Describe aquí brevemente tu planificación antes de construir el dashboard. 

- Para estructurar con éxito tu dashboard ejecutivo enfocado en la evolución 2024-2025, rentabilidad y análisis de mercado, utilizaremos un diseño de pirámide invertida (de lo general a lo particular) dividido en dos vistas estratégicas: Overview (Resumen) y Detalle.
- Vista 1: Preguntas que responde overview
    - ¿El negocio está creciendo o se está contrayendo en ingresos y utilidad respecto al año anterior?
    - ¿Qué regiones o segmentos están sosteniendo el negocio y cuáles representan un riesgo?
    - ¿Existen variaciones estacionales críticas que requieran atención?
- Vista 2: Preguntas que responde Detalle
    - ¿Cuál es el comportamiento del precio promedio y costo unitario por cada país específico?
    - ¿Qué combinaciones de segmento cliente y pais generan mayores ganancias y pérdidas inferiores al promedio?
    - ¿Qué meses o estaciones del año concentran los picos de venta y utilidad?
      
#### Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?
Especifica qué KPIs incluirás y por qué: (3-4 KPIs)
- Ingresos Totales: Muestra el tamaño bruto del negocio. 
- Utilidad Bruta: El dinero real que queda tras restar los costos.
- Margen Bruto: Mide la eficiencia operativa (Utilidad / Ingresos).
- Unidades Vendidas: Muestra el volumen del mercado. 

Explica qué gráfico usarás para cada pregunta y justifica la elección.

- Gráfico de tiempo: Gráfico de Líneas (Ingresos y Utilidad Bruta en el eje Y, Fecha pedido en el eje X). Es el estándar para identificar tendencias, pendientes de crecimiento y comparar visualmente el comportamiento de 2024 contra 2025.
- Gráfico comparación geográfica:  Gráfico de Barras Horizontal (Ingresos en el eje x, pais en el eje y, ordenado de mayor a menor). El ojo humano compara longitudes mucho más rápido que mapas o pasteles.
- Gráfico comparación clientes: Gráfico de Columnas Agrupadas (segmento cliente en el eje X, Ingresos y Utilidad Bruta como series de columnas). Permite contrastar de inmediato qué segmentos tienen un volumen alto de ventas pero una utilidad baja.


Descripción de la jerarquía visual:
- Título y Filtros (Encabezado Superior): Nombre claro del reporte a la izquierda; el selector de año a la derecha de forma discreta.
- Fila de KPIs (Franja Superior Alta): Las 4 tarjetas numéricas alineadas horizontalmente. Es lo primero que lee el director.
- Tendencia Temporal (Bloque Central - Principal): El gráfico de líneas ocupando el área de mayor atención.
- Distribución comparacion geografica y Comercial (Bloque  Inferior): Gráficos de barras de clientes .

### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 🖥️ Vista General (Overview)

**S (Situación):**  
El negocio es estable y rentable gracias al empuje de dos países clave y a los clientes de mayor valor.

**C (Complicación):**  
Los ingresos sufren caídas muy fuertes y repetitivas a mitad de año. El negocio depende demasiado de las fechas festivas de diciembre.

**Q (Pregunta):**   
¿Por qué bajan tanto las ventas a mitad de año y cómo podemos evitarlo?

**A (Respuesta):**  
Los mercados y clientes más chicos aportan muy poco. Se necesitan campañas especiales para ellos durante los meses bajos para equilibrar el año.

---

#### 🔎 Vista Detalle

**S (Situación):**  
Los segmentos más altos (Premium y Estándar) lideran la utilidad global en casi todas las regiones. Además, la temporada de verano es, por mucho, la época más fuerte y rentable del año para el negocio. (Recordar que en sudamerica verano ocurre en los meses finales del año).

**C (Complicación):**  
El segmento Económico aporta un valor casi invisible en todos los países. Al mismo tiempo, las temporadas de otoño y primavera muestran un rendimiento sumamente bajo, dejando al negocio estancado fuera del verano.

**Q (Pregunta):**  
¿Por qué el segmento Económico no despega en ninguna región y cómo podemos levantar las ventas durante las estaciones débiles del año?

**A (Respuesta):**  
La matriz operativa detalla que la baja rentabilidad fuera de temporada se debe a un bajo volumen de unidades vendidas y no a un problema de costos. La solución está en usar el segmento Económico para lanzar promociones masivas exclusivamente durante otoño y primavera para reactivar el flujo.

Sufrimos un impacto financiero recurrente por una fuerte caída de ventas a mitad de año, donde el bajo volumen de unidades vendidas frena el negocio, sumado a que el segmento Económico no está aportando valor en ninguna región.
